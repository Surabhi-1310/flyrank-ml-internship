## 3. Quick look at the data (2-3 real numbers)

*The starter dataset contains 30,000 pages and 44 columns. In the first analysis, search volume and 90-day impressions had a correlation of only 0.001, suggesting that search volume alone is not a strong indicator of page exposure in this sample. Mean CTR also varied from 0.3548 for page-one results to 0.0554 for deep results, showing that position is an important context when reviewing CTR.*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("Search volume vs impressions correlation:",
      round(df["search_volume"].corr(df["impressions_90d"]), 3))

visible = df[df["impressions_90d"] >= 100]
print("Page 1 mean CTR:",
      round(visible.loc[visible["position_tier"] == "page_1", "ctr"].mean(), 4))
print("Deep mean CTR:",
      round(visible.loc[visible["position_tier"] == "deep", "ctr"].mean(), 4))
```
